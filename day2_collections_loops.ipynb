```python
Numbers = [1,2,3,4]
for n in Numbers:
    print(n*n)
```

```python
Numbers=[7,6,8,9]
for index, n in enumerate(Numbers):
    print(index*n)
def evenList(fullList):
    even = []
    for n in fullList:
        if(n%2 == 0):
            even.append(n)
    return even
print(evenList(Numbers))
```
